```python
import os
import sys
import json
import pandas as pd
from pathlib import Path

# Set working directory
PROJECT_ROOT = Path(r"C:\Repos\codecritic").resolve()
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))

# Load log CSV
BACKUP_PATH = Path("tests/backup/provider_logs.csv")
assert BACKUP_PATH.exists(), "❌ Backup file not found"
df = pd.read_csv(BACKUP_PATH)

# Filter for agent_engine provider rows
df_engine = df[df["provider_type"] == "PROVIDER_TYPE.AGENT_ENGINE"].copy()

# Parse output assuming AgentEngineOutput schema
def parse_output(row):
    raw = row.get("output")

    if isinstance(raw, float) and pd.isna(raw):
        return pd.Series({})
    if isinstance(raw, dict):
        parsed = raw
    elif isinstance(raw, str):
        try:
            parsed = json.loads(raw)
        except json.JSONDecodeError:
            return pd.Series({})
    else:
        return pd.Series({})

    return pd.Series({
        "response": parsed.get("response"),
        "agent_decision": parsed.get("agent_decision"),
        "conversation_log_entry": parsed.get("conversation_log_entry")
    })

# Expand output and combine with original DataFrame
parsed_output = df_engine.apply(parse_output, axis=1)
df_engine_expanded = pd.concat([df_engine.reset_index(drop=True), parsed_output.reset_index(drop=True)], axis=1)

# Show result
df_engine_expanded
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

# Step 1: Copy the dataframe for safe manipulation
df_latency = df_engine_expanded.copy()

# Step 2: Map provider_id → model
provider_id_to_model = {
    1: "mock-llm",
    2: "gpt-4o",
    3: "gemini-1.5-pro",
    4: "claude-3-sonnet",
}
df_latency['model'] = df_latency['provider_id'].map(provider_id_to_model)
df_latency.to_csv('temp.csv')
# Step 3: Drop rows missing model or latency
df_latency = df_latency.dropna(subset=['model', 'latency_ms'])

# Step 4: Filter out models with only zero/invalid latency
valid_models = (
    df_latency
    .groupby("model")["latency_ms"]
    .apply(lambda x: (x > 0).sum())
    .loc[lambda s: s > 0]
    .index.tolist()
)
df_latency = df_latency[df_latency["model"].isin(valid_models)]

# Step 5: Assign numeric x-coordinates and jitter
model_names = sorted(df_latency["model"].unique())
model_to_num = {model: i for i, model in enumerate(model_names)}
df_latency["x_numeric"] = df_latency["model"].map(model_to_num)
df_latency["x_jittered"] = df_latency["x_numeric"] + np.random.uniform(-0.25, 0.25, size=len(df_latency))

# Step 6: Create box traces
box_traces = []
for model in model_names:
    model_data = df_latency[df_latency["model"] == model]
    box_traces.append(go.Box(
        y=model_data["latency_ms"],
        x=[model_to_num[model]] * len(model_data),
        name=model,
        boxpoints=False,
        fillcolor='rgba(120, 81, 169, 0.25)',
        line_color='rgba(80, 50, 120, 0.9)',
        marker_color='rgba(80, 50, 120, 0.9)',
        width=0.5,
        showlegend=False
    ))

# Step 7: Jittered scatter trace
scatter_trace = go.Scatter(
    x=df_latency["x_jittered"],
    y=df_latency["latency_ms"],
    mode='markers',
    marker=dict(
        color='rgba(120, 81, 169, 0.4)',
        size=5
    ),
    hoverinfo='skip',
    showlegend=False
)

# Step 8: Final layout
fig = go.Figure(data=box_traces + [scatter_trace])
fig.update_layout(
    title="Latency Distribution by Agent Engine",
    xaxis=dict(
        tickmode='array',
        tickvals=list(model_to_num.values()),
        ticktext=list(model_to_num.keys()),
        tickangle=20,
        title="Agent Engine"
    ),
    yaxis_title="Latency (ms)",
    width=1500,
    height=500,
    plot_bgcolor='white',
    margin=dict(t=60, b=80)
)

fig.show()
```

```python
import pandas as pd
import json
import numpy as np
import plotly.graph_objects as go

# Step 1: Copy for safe transformation
df_cost = df_engine_expanded.copy()

# Step 2: Extract real prompt and response from JSON
def extract_prompt_response(row):
    try:
        input_data = json.loads(row["input"]) if isinstance(row["input"], str) else {}
        output_data = json.loads(row["output"]) if isinstance(row["output"], str) else {}

        prompt_text = input_data.get("prompt", {}).get("prompt", "")
        response_text = output_data.get("response", "")

        return pd.Series({
            "input_text": prompt_text,
            "output_text": response_text
        })
    except Exception:
        return pd.Series({"input_text": "", "output_text": ""})

# Apply extraction
df_cost[["input_text", "output_text"]] = df_cost.apply(extract_prompt_response, axis=1)

# Step 3: Estimate tokens from words (input + output)
df_cost["input_words"] = df_cost["input_text"].str.split().apply(len)
df_cost["output_words"] = df_cost["output_text"].str.split().apply(len)
df_cost["total_words"] = df_cost["input_words"] + df_cost["output_words"]
df_cost["approx_tokens"] = df_cost["total_words"] * 1.4  # conservative multiplier

# Step 4: Map provider → model and token cost
provider_id_to_model = {
    1: "mock-llm",
    2: "gpt-4o",
    3: "gemini-1.5-pro",
    4: "claude-3-sonnet",
}
cost_map = {
    "mock-llm": 0.0,
    "gpt-4o": 0.005,
    "gemini-1.5-pro": 0.007,
    "claude-3-sonnet": 0.003,
}

df_cost["model"] = df_cost["provider_id"].map(provider_id_to_model)
df_cost["cost_per_1k_tokens"] = df_cost["model"].map(cost_map)
df_cost["estimated_cost"] = (df_cost["approx_tokens"] / 1000) * df_cost["cost_per_1k_tokens"]

# Step 5: Summarize by model
# Step 5: Summarize by model, including token count
summary = (
    df_cost
    .dropna(subset=["estimated_cost", "model"])
    .groupby("model")
    .agg(
        run_count=("id", "count"),
        total_tokens=("approx_tokens", "sum"),
        total_cost=("estimated_cost", "sum"),
        avg_cost=("estimated_cost", "mean")
    )
    .reset_index()
)

# Step 6: Plot
fig = go.Figure()
fig.add_trace(go.Bar(
    x=summary["model"],
    y=summary["total_cost"],
    text=summary["run_count"],
    textposition="outside",
    marker_color="rgba(120, 81, 169, 0.6)"
))
fig.update_layout(
    title="Estimated Total Cost by Agent Engine (Based on Prompt + Response Token Count)",
    xaxis_title="Agent Engine Model",
    yaxis_title="Total Cost (USD)",
    width=1500,
    height=450,
    plot_bgcolor="white",
    margin=dict(t=60, b=60),
    showlegend=False
)

fig.show()
```

```python
from collections import Counter
import plotly.express as px

# Step 1: Copy and prep
df_words = df_engine_expanded.copy()
log_texts = df_words["conversation_log_entry"].fillna("").str.lower().tolist()

# Step 2: Simple keyword tokenization (split on whitespace, basic cleanup)
def tokenize(text):
    return [word.strip(".,:;()[]") for word in text.split() if word.isalpha() and len(word) > 3]

all_words = []
for entry in log_texts:
    all_words.extend(tokenize(entry))

# Step 3: Count and get top N
top_n = 20
word_counts = Counter(all_words).most_common(top_n)
keywords, counts = zip(*word_counts)

# Step 4: Plot
fig = px.bar(
    x=keywords,
    y=counts,
    title="Top Keywords in Conversation Logs",
    labels={"x": "Keyword", "y": "Frequency"},
    text=counts
)

fig.update_traces(marker_color="rgba(120, 81, 169, 0.6)", textposition="outside")
fig.update_layout(
    xaxis_title="Keyword",
    yaxis_title="Count",
    width=1000,
    height=450,
    plot_bgcolor="white",
    margin=dict(t=60, b=60)
)

fig.show()
```

```python
import plotly.graph_objects as go
import json
from Levenshtein import distance as levenshtein_distance

# Step 1: Extract before/after code
df_table = df_engine_expanded.copy()

import re

def extract_code_pair(row):
    try:
        # Load input/output safely
        input_data = json.loads(row["input"]) if isinstance(row["input"], str) else {}
        prompt_block = input_data.get("prompt", {})
        prompt_text = prompt_block.get("prompt", "")

        # Extract source_code using regex from within prompt string
        match = re.search(r'"source_code"\s*:\s*"((?:[^"\\]|\\.)*)"', prompt_text)
        before = match.group(1) if match else ""

        # Extract [CODE] block from output
        output_data = json.loads(row["output"]) if isinstance(row["output"], str) else {}
        after = output_data.get("response", "")
        if "[CODE]" in after and "[/CODE]" in after:
            after = after.split("[CODE]", 1)[1].split("[/CODE]", 1)[0].strip()

        return pd.Series({
            "before_code": before.strip(),
            "after_code": after.strip()
        })
    except Exception as e:
        print(f"❌ Failed on row: {e}")
        return pd.Series({"before_code": "", "after_code": ""})


df_table[["before_code", "after_code"]] = df_table.apply(extract_code_pair, axis=1)

# Step 2: Compute Levenshtein distance
df_table["code_diff"] = df_table.apply(
    lambda row: levenshtein_distance(row["before_code"], row["after_code"]),
    axis=1
)

# Step 3: Prepare top N for display
df_top = df_table[df_table["code_diff"] > 0].copy()
df_top = df_top.sort_values("code_diff", ascending=False).head(20)

# Truncate code for table readability
df_top["before_preview"] = df_top["before_code"].str.slice(0, 100).str.replace("\n", " ↵ ")
df_top["after_preview"] = df_top["after_code"].str.slice(0, 100).str.replace("\n", " ↵ ")

# Step 4: Build Plotly table
fig = go.Figure(data=[go.Table(
    header=dict(
        values=["File", "Levenshtein Distance", "Before (truncated)", "After (truncated)"],
        fill_color="rgba(120, 81, 169, 0.2)",
        line_color="darkslategray",
        align="left"
    ),
    cells=dict(
        values=[
            df_top["file_name"],
            df_top["code_diff"],
            df_top["before_preview"],
            df_top["after_preview"]
        ],
        fill_color="white",
        line_color="lightgray",
        align="left",
        height=28
    )
)])

fig.update_layout(
    title="Top Code Changes – Side-by-Side Comparison",
    width=1200,
    height=800,
    margin=dict(t=80, b=60)
)

fig.show()
```

```python
import plotly.express as px
import json
from Levenshtein import distance as levenshtein_distance
import re

# Step 1: Copy and extract
df_percent = df_engine_expanded.copy()

def extract_code_pair(row):
    try:
        input_data = json.loads(row["input"]) if isinstance(row["input"], str) else {}
        prompt_block = input_data.get("prompt", {})
        prompt_text = prompt_block.get("prompt", "")

        match = re.search(r'"source_code"\s*:\s*"((?:[^"\\]|\\.)*)"', prompt_text)
        before = match.group(1) if match else ""

        output_data = json.loads(row["output"]) if isinstance(row["output"], str) else {}
        after = output_data.get("response", "")
        if "[CODE]" in after and "[/CODE]" in after:
            after = after.split("[CODE]", 1)[1].split("[/CODE]", 1)[0].strip()

        return pd.Series({
            "before_code": before.strip(),
            "after_code": after.strip()
        })
    except Exception:
        return pd.Series({"before_code": "", "after_code": ""})

df_percent[["before_code", "after_code"]] = df_percent.apply(extract_code_pair, axis=1)

# Step 2: Levenshtein + percent change
df_percent["code_diff"] = df_percent.apply(
    lambda row: levenshtein_distance(row["before_code"], row["after_code"]),
    axis=1
)
df_percent["before_len"] = df_percent["before_code"].str.len().clip(lower=1)  # avoid division by 0
df_percent["percent_change"] = (df_percent["code_diff"] / df_percent["before_len"]) * 100

# Step 3: Filter and label
df_percent = df_percent[df_percent["percent_change"] > 0].copy()
df_percent = df_percent.sort_values("percent_change", ascending=False).reset_index(drop=True)
df_percent["label"] = df_percent["file_name"].fillna("").apply(lambda f: f.split("\\")[-1])  # or use index

# Step 4: Plot
fig = px.bar(
    df_percent.head(30),
    x="label",
    y="percent_change",
    hover_data=["before_code", "after_code"],
    labels={
        "label": "Run / File",
        "percent_change": "Percent Code Change"
    },
    title="Normalized Code Change (Levenshtein %)"
)

fig.update_layout(
    xaxis_tickangle=45,
    width=1100,
    height=500,
    margin=dict(t=60, b=100),
    plot_bgcolor="white"
)

fig.show()
```
