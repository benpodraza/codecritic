# 📒 Notebook: Symbol Library Validation and Exploration

```

import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parents[2]
sys.path.insert(0, str(PROJECT_ROOT))

from app.utilities.symbol_library import SymbolGraph
        
```

## Prepare Test Files

```

test_dir = PROJECT_ROOT / "tests" / "symbol_tests"
test_dir.mkdir(parents=True, exist_ok=True)

(test_dir / "test_simple.py").write_text('''
def greet(name):
    print(f"Hello, {name}")
''', encoding="utf-8")

(test_dir / "test_complex.py").write_text('''
class Calculator:
    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b

def helper():
    pass
''', encoding="utf-8")

print("✅ Test files created.")
        
```

## Validate Basic Symbol Extraction

```

symbol_graph = SymbolGraph()
symbol_graph.parse_file(test_dir / "test_simple.py")

print("✅ Symbols from test_simple.py:")
for symbol in symbol_graph.symbols:
    print(symbol)
        
```

## Validate Complex Symbol Extraction

```

symbol_graph = SymbolGraph()
symbol_graph.parse_file(test_dir / "test_complex.py")

print("✅ Symbols from test_complex.py:")
for symbol in symbol_graph.symbols:
    print(symbol)
        
```

## Validate Graph Structure

```

print("✅ Symbol Graph Structure for test_complex.py:")
for node, children in symbol_graph.graph.items():
    print(f"{node}: {children}")
        
```

## Test Edge Cases

```

empty_test_file = test_dir / "empty.py"
empty_test_file.touch()

symbol_graph = SymbolGraph()
symbol_graph.parse_file(empty_test_file)

print("✅ Symbols from empty.py (should be empty):")
print(symbol_graph.symbols)

non_existent_file = test_dir / "does_not_exist.py"
try:
    symbol_graph.parse_file(non_existent_file)
except Exception as e:
    print("✅ Caught expected error for non-existent file:", e)
        
```

## Internals Inspection

```

print("✅ Internal symbol representation:")
for symbol in symbol_graph.symbols:
    print("Symbol Name:", symbol.name)
    print("Symbol Type:", symbol.type)
    print("Defined at:", symbol.file_path, "Line:", symbol.line_number)
        
```
