```python
# 🧪 PromptProvider Log Expansion — Full Record Enrichment with Prompt Output

import os, sys, json
import pandas as pd
from pathlib import Path

# 📁 Set up paths
PROJECT_ROOT = Path(r"C:\Repos\codecritic").resolve()
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))

LOG_PATH = Path("tests/backup/provider_logs.csv")
assert LOG_PATH.exists(), "❌ provider_logs.csv not found"

# 📥 Load log data
df = pd.read_csv(LOG_PATH)

# 🔍 Filter only PROMPT provider logs
df_prompt = df[df["provider_type"] == "PROVIDER_TYPE.PROMPT"].copy()

# 🧼 Safe JSON parse for output column
def safe_json_parse(val):
    try:
        return json.loads(val) if isinstance(val, str) else {}
    except Exception:
        return {}

df_prompt["output_parsed"] = df_prompt["output"].apply(safe_json_parse)

# 🧩 Extract PromptOutputSchema fields into new columns
df_prompt["prompt"] = df_prompt["output_parsed"].apply(lambda x: x.get("prompt"))
df_prompt["summary"] = df_prompt["output_parsed"].apply(lambda x: x.get("summary"))

# 📐 Preview enriched full records
pd.set_option("display.max_colwidth", 200)
df_prompt.sort_values("timestamp").head(20)
```
