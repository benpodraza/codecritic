```python
# 🧪 AgentEngineProvider Log Expansion — Full Schema Alignment

import os, sys, json
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path(r"C:\Repos\codecritic").resolve()
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))

LOG_PATH = Path("tests/backup/provider_logs.csv")
assert LOG_PATH.exists(), "❌ provider_logs.csv not found"

df = pd.read_csv(LOG_PATH)
df_agent = df[df["provider_type"] == "PROVIDER_TYPE.AGENT_ENGINE"].copy()

def safe_json_parse(val):
    try:
        return json.loads(val) if isinstance(val, str) else {}
    except Exception:
        return {}

# Parse JSON output field
df_agent["output_parsed"] = df_agent["output"].apply(safe_json_parse)

# Extract all AgentEngineOutput fields
df_agent["response"] = df_agent["output_parsed"].apply(lambda x: x.get("response"))
df_agent["token_count"] = df_agent["output_parsed"].apply(lambda x: x.get("token_count"))
df_agent["cost_usd"] = df_agent["output_parsed"].apply(lambda x: x.get("cost_usd"))
df_agent["snapshot_id"] = df_agent["output_parsed"].apply(lambda x: x.get("snapshot_id"))
df_agent["summary"] = df_agent["output_parsed"].apply(lambda x: x.get("summary"))
df_agent["code"] = df_agent["output_parsed"].apply(lambda x: x.get("code"))
df_agent["decision"] = df_agent["output_parsed"].apply(lambda x: x.get("decision"))
df_agent["conversation_log_entry"] = df_agent["output_parsed"].apply(lambda x: x.get("conversation_log_entry"))

# Drop intermediate column
df_agent_expanded = df_agent.drop(columns=["output_parsed"])

# Show only relevant columns (schema-aligned)
column_order = [
    # ProviderLogSchema fields
    "timestamp", "session_id", "provider_id", "provider_type", "file_log_id", "run_id",
    "called_by_type", "called_by_id", "parent_id", "execution_chain", "file_name", 
    "latency_ms", "output_schema", "config_hash", "input", "output",
    
    # AgentEngineOutput fields
    "decision", "summary", "conversation_log_entry",
    "response", "code", "token_count", "cost_usd", "snapshot_id"
]

df_agent_expanded = df_agent_expanded[column_order]
pd.set_option("display.max_colwidth", 200)
df_agent_expanded.sort_values("timestamp").head(20)
```

```python
import plotly.express as px

fig_latency = px.bar(
    df_agent,
    x="file_log_id",
    y="latency_ms",
    hover_data=["timestamp", "run_id"],
    title="⏱️ Latency per AgentEngine Call",
    labels={"file_log_id": "File Log ID", "latency_ms": "Latency (ms)"}
)
fig_latency.show()

```

```python
fig_cost = px.bar(
    df_agent,
    x="file_log_id",
    y="cost_usd",
    hover_data=["token_count", "timestamp"],
    title="💵 Cost per AgentEngine Call",
    labels={"file_log_id": "File Log ID", "cost_usd": "Cost (USD)"}
)
fig_cost.show()
```
